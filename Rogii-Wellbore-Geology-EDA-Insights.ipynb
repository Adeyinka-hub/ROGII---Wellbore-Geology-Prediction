```python
%%time
# =============================================================================
# 1. SETUP & CONFIGURATION
# =============================================================================
import os
import sys
import warnings
from pathlib import Path
from collections import Counter, defaultdict
from typing import List, Dict, Tuple, Optional
import math

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.signal import correlate, find_peaks, savgol_filter
from scipy.interpolate import interp1d, UnivariateSpline
from scipy.spatial.distance import cdist
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Configuration
warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', 100)
pd.set_option('display.max_rows', 200)
pd.set_option('display.max_colwidth', 200)
sns.set_theme(style='whitegrid', context='notebook')

# Random state for reproducibility
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

# Colors for consistent visualization
COLORS = {
    'primary': '#2E86AB',
    'secondary': '#A23B72',
    'accent': '#F18F01',
    'success': '#28a745',
    'warning': '#ffc107',
    'danger': '#dc3545',
    'neutral': '#6c757d',
    'gr': '#1f77b4',
    'tvt': '#ff7f0e',
    'spatial': '#2ca02c',
    'geology': '#9467bd'
}

print("✅ Setup complete")

# %%
# =============================================================================
# 2. DATA LOADING & STRUCTURE VALIDATION
# =============================================================================

class WellboreDataLoader:
    """Efficient, memory-conscious loader for wellbore competition data"""

    def __init__(self, data_root: str):
        self.data_root = Path(data_root)
        self.train_dir = self.data_root / 'train'
        self.test_dir = self.data_root / 'test'
        self._train_wells = None
        self._test_wells = None

    def _get_well_ids(self, directory: Path) -> List[str]:
        """Extract well IDs from filename pattern"""
        pattern = '*__horizontal_well.csv'
        files = list(directory.glob(pattern))
        return sorted([f.stem.split('__')[0] for f in files])

    @property
    def train_well_ids(self) -> List[str]:
        if self._train_wells is None:
            self._train_wells = self._get_well_ids(self.train_dir)
        return self._train_wells

    @property
    def test_well_ids(self) -> List[str]:
        if self._test_wells is None:
            self._test_wells = self._get_well_ids(self.test_dir)
        return self._test_wells

    def load_horizontal(self, well_id: str, split: str = 'train',
                       usecols: Optional[List[str]] = None) -> pd.DataFrame:
        """Load horizontal well CSV with optional column selection"""
        directory = self.train_dir if split == 'train' else self.test_dir
        filepath = directory / f'{well_id}__horizontal_well.csv'

        df = pd.read_csv(filepath, usecols=usecols)
        df['well_id'] = well_id
        df['split'] = split
        return df

    def load_typewell(self, well_id: str, split: str = 'train',
                     usecols: Optional[List[str]] = None) -> Optional[pd.DataFrame]:
        """Load typewell CSV; returns None if not found"""
        directory = self.train_dir if split == 'train' else self.test_dir
        filepath = directory / f'{well_id}__typewell.csv'

        if not filepath.exists():
            return None
        df = pd.read_csv(filepath, usecols=usecols)
        df['well_id'] = well_id
        df['split'] = split
        return df

    def load_well_pair(self, well_id: str, split: str = 'train') -> Tuple[pd.DataFrame, Optional[pd.DataFrame]]:
        """Load both horizontal and typewell for a well"""
        return self.load_horizontal(well_id, split), self.load_typewell(well_id, split)

    def get_file_inventory(self) -> pd.DataFrame:
        """Create inventory of all files in dataset"""
        inventory = []

        for split_dir, split_name in [(self.train_dir, 'train'), (self.test_dir, 'test')]:
            if not split_dir.exists():
                continue
            for filepath in split_dir.iterdir():
                if filepath.suffix in ['.csv', '.png', '.pptx']:
                    inventory.append({
                        'split': split_name,
                        'filename': filepath.name,
                        'well_id': filepath.stem.split('__')[0] if '__' in filepath.stem else None,
                        'file_type': filepath.suffix.lstrip('.'),
                        'size_mb': filepath.stat().st_size / (1024**2)
                    })

        # Add root files
        for filepath in self.data_root.glob('*'):
            if filepath.suffix in ['.csv', '.pptx'] and filepath.name not in ['train', 'test']:
                inventory.append({
                    'split': 'root',
                    'filename': filepath.name,
                    'well_id': None,
                    'file_type': filepath.suffix.lstrip('.'),
                    'size_mb': filepath.stat().st_size / (1024**2)
                })

        return pd.DataFrame(inventory)

# Initialize loader
DATA_ROOT = Path('/kaggle/input/competitions/rogii-wellbore-geology-prediction')
loader = WellboreDataLoader(DATA_ROOT)

print(f"📊 Dataset Inventory:")
inventory_df = loader.get_file_inventory()
display(inventory_df.groupby(['split', 'file_type']).agg({
    'filename': 'count',
    'size_mb': 'sum'
}).round(2).rename(columns={'filename': 'count', 'size_mb': 'total_mb'}))

print(f"\n🔢 Well counts:")
print(f"  Training wells: {len(loader.train_well_ids)}")
print(f"  Test wells (visible): {len(loader.test_well_ids)}")

# %%
# =============================================================================
# 3. HORIZONTAL WELL DEEP DIVE
# =============================================================================

def analyze_horizontal_well(well_id: str, split: str = 'train') -> Dict:
    """Comprehensive analysis of a single horizontal well"""
    horizontal, typewell = loader.load_well_pair(well_id, split)

    analysis = {
        'well_id': well_id,
        'split': split,
        'horizontal_shape': horizontal.shape,
        'typewell_shape': typewell.shape if typewell is not None else None,
        'columns': list(horizontal.columns),
        'dtypes': horizontal.dtypes.to_dict(),
        'memory_mb': horizontal.memory_usage(deep=True).sum() / (1024**2),
        'missing_summary': horizontal.isnull().sum().to_dict(),
        'numeric_stats': horizontal.select_dtypes(include=[np.number]).describe().to_dict(),
    }

    # TVT-specific analysis
    if 'TVT' in horizontal.columns:
        tvt_known = horizontal['TVT'].notna().sum()
        tvt_missing = horizontal['TVT'].isna().sum()
        analysis['tvt_coverage'] = {
            'known': int(tvt_known),
            'missing': int(tvt_missing),
            'coverage_pct': round(100 * tvt_known / len(horizontal), 2)
        }

    # TVT_input analysis (critical for prediction zone)
    if 'TVT_input' in horizontal.columns:
        input_known = horizontal['TVT_input'].notna().sum()
        input_missing = horizontal['TVT_input'].isna().sum()
        first_missing_idx = horizontal['TVT_input'].isna().idxmax() if input_missing > 0 else None

        analysis['tvt_input'] = {
            'known_rows': int(input_known),
            'missing_rows': int(input_missing),
            'missing_pct': round(100 * input_missing / len(horizontal), 2),
            'first_missing_index': int(first_missing_idx) if first_missing_idx is not None else None,
            'prediction_start_md': float(horizontal.loc[first_missing_idx, 'MD']) if first_missing_idx is not None else None
        }

    # GR analysis
    if 'GR' in horizontal.columns:
        gr_valid = horizontal['GR'].dropna()
        analysis['gr_stats'] = {
            'min': float(gr_valid.min()),
            'max': float(gr_valid.max()),
            'mean': float(gr_valid.mean()),
            'std': float(gr_valid.std()),
            'missing_pct': round(100 * horizontal['GR'].isna().sum() / len(horizontal), 2)
        }

    # Spatial extent
    if all(col in horizontal.columns for col in ['X', 'Y', 'Z']):
        analysis['spatial_extent'] = {
            'X_range': float(horizontal['X'].max() - horizontal['X'].min()),
            'Y_range': float(horizontal['Y'].max() - horizontal['Y'].min()),
            'Z_range': float(horizontal['Z'].max() - horizontal['Z'].min()),
            'total_md': float(horizontal['MD'].max() - horizontal['MD'].min())
        }

    return analysis

# Sample wells for detailed inspection
sample_wells = loader.train_well_ids[:5]
print(f"\n🔍 Analyzing sample wells: {sample_wells}")

well_analyses = [analyze_horizontal_well(wid) for wid in sample_wells]
sample_summary = pd.DataFrame([{
    'well_id': a['well_id'],
    'rows': a['horizontal_shape'][0],
    'TVT_known': a.get('tvt_coverage', {}).get('known', 'N/A'),
    'TVT_input_known': a.get('tvt_input', {}).get('known_rows', 'N/A'),
    'prediction_start_idx': a.get('tvt_input', {}).get('first_missing_index', 'N/A'),
    'GR_missing_pct': a.get('gr_stats', {}).get('missing_pct', 'N/A'),
    'MD_range_ft': a.get('spatial_extent', {}).get('total_md', 'N/A')
} for a in well_analyses])

display(sample_summary)

# %%
# =============================================================================
# 4. TVT TARGET ANALYSIS & PREDICTION ZONE CHARACTERIZATION
# =============================================================================

def extract_prediction_zone_info(horizontal: pd.DataFrame) -> Dict:
    """Extract key information about the prediction zone structure"""
    if 'TVT_input' not in horizontal.columns:
        return {}

    # Find prediction start (first NaN in TVT_input)
    is_known = horizontal['TVT_input'].notna()

    if not is_known.any() or not (~is_known).any():
        return {'error': 'No clear prediction boundary'}

    # Find transition point
    transition_idx = is_known.astype(int).diff().idxmin()

    return {
        'total_rows': len(horizontal),
        'known_rows': int(is_known.sum()),
        'prediction_rows': int((~is_known).sum()),
        'prediction_pct': round(100 * (~is_known).sum() / len(horizontal), 2),
        'transition_index': int(transition_idx),
        'transition_md': float(horizontal.loc[transition_idx, 'MD']) if transition_idx in horizontal.index else None,
        'known_md_range': (
            float(horizontal.loc[is_known, 'MD'].min()),
            float(horizontal.loc[is_known, 'MD'].max())
        ),
        'prediction_md_range': (
            float(horizontal.loc[~is_known, 'MD'].min()) if (~is_known).any() else None,
            float(horizontal.loc[~is_known, 'MD'].max()) if (~is_known).any() else None
        )
    }

# Analyze prediction zone across all training wells
print("📐 Prediction Zone Analysis (Training Set)")
prediction_stats = []

for well_id in loader.train_well_ids:
    horizontal = loader.load_horizontal(well_id, 'train')
    zone_info = extract_prediction_zone_info(horizontal)
    if zone_info and 'error' not in zone_info:
        zone_info['well_id'] = well_id
        prediction_stats.append(zone_info)

pred_df = pd.DataFrame(prediction_stats)

if len(pred_df) > 0:
    display(pred_df[['well_id', 'total_rows', 'known_rows', 'prediction_rows', 'prediction_pct']].head(10))

    print(f"\n📊 Prediction Zone Statistics:")
    print(f"  Mean prediction %: {pred_df['prediction_pct'].mean():.1f}%")
    print(f"  Std prediction %: {pred_df['prediction_pct'].std():.1f}%")
    print(f"  Min prediction %: {pred_df['prediction_pct'].min():.1f}%")
    print(f"  Max prediction %: {pred_df['prediction_pct'].max():.1f}%")

    # Visualization
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    # Distribution of prediction percentages
    axes[0, 0].hist(pred_df['prediction_pct'], bins=30, edgecolor='white', color=COLORS['primary'])
    axes[0, 0].axvline(pred_df['prediction_pct'].mean(), color=COLORS['accent'], linestyle='--', linewidth=2)
    axes[0, 0].set_xlabel('Prediction Zone Percentage')
    axes[0, 0].set_ylabel('Well Count')
    axes[0, 0].set_title('Distribution of Prediction Zone Size')
    axes[0, 0].grid(alpha=0.3)

    # Known vs Prediction rows scatter
    axes[0, 1].scatter(pred_df['known_rows'], pred_df['prediction_rows'],
                      alpha=0.6, color=COLORS['secondary'], edgecolors='white')
    axes[0, 1].plot([0, pred_df['known_rows'].max()], [0, pred_df['known_rows'].max()],
                   'r--', alpha=0.5)
    axes[0, 1].set_xlabel('Known TVT_input Rows')
    axes[0, 1].set_ylabel('Prediction Zone Rows')
    axes[0, 1].set_title('Known vs Prediction Rows')
    axes[0, 1].grid(alpha=0.3)

    # Prediction % vs Total Rows
    axes[1, 0].scatter(pred_df['total_rows'], pred_df['prediction_pct'],
                      alpha=0.6, color=COLORS['success'])
    axes[1, 0].set_xlabel('Total Well Rows')
    axes[1, 0].set_ylabel('Prediction Percentage')
    axes[1, 0].set_title('Prediction % vs Well Length')
    axes[1, 0].grid(alpha=0.3)

    # Transition MD distribution
    valid_md = pred_df['transition_md'].dropna()
    if len(valid_md) > 0:
        axes[1, 1].hist(valid_md, bins=30, edgecolor='white', color=COLORS['geology'])
        axes[1, 1].set_xlabel('Transition MD (ft)')
        axes[1, 1].set_ylabel('Well Count')
        axes[1, 1].set_title('Distribution of Prediction Start Points')
        axes[1, 1].grid(alpha=0.3)

    plt.tight_layout()
    plt.show()

# %%
# =============================================================================
# 5. TVT DIRECTION & TREND ANALYSIS (Critical Insight!)
# =============================================================================

def compute_tvt_derivatives(horizontal: pd.DataFrame) -> pd.DataFrame:
    """Compute TVT rate-of-change features"""
    df = horizontal.copy()

    if 'TVT' not in df.columns or 'MD' not in df.columns:
        return df

    # First derivative: dTVT/dMD
    df['dTVT'] = df['TVT'].diff()
    df['dMD'] = df['MD'].diff()
    df['dTVT_dMD'] = df['dTVT'] / df['dMD']

    # Clean infinities
    df['dTVT_dMD'] = df['dTVT_dMD'].replace([np.inf, -np.inf], np.nan)

    # Second derivative (curvature)
    df['d2TVT_dMD2'] = df['dTVT_dMD'].diff() / df['dMD']
    df['d2TVT_dMD2'] = df['d2TVT_dMD2'].replace([np.inf, -np.inf], np.nan)

    return df

def classify_tvt_trend(dTVT_dMD: pd.Series, flat_threshold: float = 0.02) -> pd.Series:
    """Classify TVT trend direction"""
    return pd.Series(np.select(
        [dTVT_dMD > flat_threshold, dTVT_dMD < -flat_threshold],
        ['increasing', 'decreasing'],
        default='nearly_flat'
    ))

# Analyze TVT trends across training wells
print("📈 TVT Trend Analysis")
trend_samples = []

for well_id in loader.train_well_ids[:100]:  # Sample for efficiency
    horizontal = loader.load_horizontal(well_id, 'train')
    if 'TVT' not in horizontal.columns:
        continue

    df = compute_tvt_derivatives(horizontal)
    valid = df['dTVT_dMD'].dropna()

    if len(valid) < 100:
        continue

    # Sample to avoid bias from long wells
    if len(valid) > 500:
        valid = valid.sample(500, random_state=RANDOM_STATE)

    trend_samples.append(pd.DataFrame({
        'well_id': well_id,
        'dTVT_dMD': valid.values,
        'trend': classify_tvt_trend(valid).values
    }))

if trend_samples:
    trend_df = pd.concat(trend_samples, ignore_index=True)

    # Trend distribution
    trend_counts = trend_df['trend'].value_counts(normalize=True) * 100
    print(f"\nTVT Trend Distribution (sampled):")
    for trend, pct in trend_counts.items():
        print(f"  {trend}: {pct:.1f}%")

    # Visualization
    fig, axes = plt.subplots(1, 3, figsize=(16, 5))

    # Histogram of dTVT/dMD
    axes[0].hist(trend_df['dTVT_dMD'].clip(-2, 2), bins=80,
                edgecolor='white', color=COLORS['primary'], alpha=0.8)
    axes[0].axvline(0, color='black', linewidth=1, linestyle='--')
    axes[0].set_xlabel('dTVT/dMD')
    axes[0].set_ylabel('Count')
    axes[0].set_title('TVT Rate of Change Distribution (clipped)')
    axes[0].grid(alpha=0.3)

    # Trend pie chart
    axes[1].pie(trend_counts, labels=trend_counts.index, autopct='%1.1f%%',
               colors=[COLORS['success'], COLORS['danger'], COLORS['neutral']])
    axes[1].set_title('TVT Trend Classification')

    # Trend by well (sample wells)
    sample_wells_trend = trend_df['well_id'].value_counts().head(10).index
    well_trends = trend_df[trend_df['well_id'].isin(sample_wells_trend)].groupby(
        ['well_id', 'trend']).size().unstack(fill_value=0)

    well_trends.plot(kind='barh', stacked=True, ax=axes[2],
                    color=[COLORS['success'], COLORS['neutral'], COLORS['danger']])
    axes[2].set_xlabel('Sample Count')
    axes[2].set_title('Trend Distribution by Well (Top 10)')
    axes[2].legend(loc='lower right')
    axes[2].grid(alpha=0.3, axis='x')

    plt.tight_layout()
    plt.show()

    # Key insight: TVT is NOT monotonic!
    print(f"\n⚠️  KEY INSIGHT: TVT can increase, decrease, or stay flat!")
    print(f"   Models assuming monotonic TVT will fail on ~48% of segments.")

# %%
# =============================================================================
# 6. GAMMA RAY (GR) SIGNAL ANALYSIS
# =============================================================================

def analyze_gr_signal(horizontal: pd.DataFrame, typewell: Optional[pd.DataFrame],
                     well_id: str) -> Dict:
    """Analyze GR signal characteristics and quality"""
    analysis = {'well_id': well_id}

    # Horizontal GR stats
    if 'GR' in horizontal.columns:
        gr_h = horizontal['GR'].dropna()
        analysis['horizontal_gr'] = {
            'count': len(gr_h),
            'missing_pct': 100 * horizontal['GR'].isna().sum() / len(horizontal),
            'min': float(gr_h.min()),
            'max': float(gr_h.max()),
            'mean': float(gr_h.mean()),
            'std': float(gr_h.std()),
            'median': float(gr_h.median()),
            'iqr': float(gr_h.quantile(0.75) - gr_h.quantile(0.25)),
            'skewness': float(stats.skew(gr_h)),
            'kurtosis': float(stats.kurtosis(gr_h))
        }

    # Typewell GR stats
    if typewell is not None and 'GR' in typewell.columns:
        gr_t = typewell['GR'].dropna()
        analysis['typewell_gr'] = {
            'count': len(gr_t),
            'min': float(gr_t.min()),
            'max': float(gr_t.max()),
            'mean': float(gr_t.mean()),
            'std': float(gr_t.std()),
            'resolution_ft': float(typewell['TVT'].diff().median()) if 'TVT' in typewell.columns else None
        }

    # GR signal quality metrics
    if 'GR' in horizontal.columns and len(horizontal) > 10:
        gr_clean = horizontal['GR'].dropna()
        if len(gr_clean) > 10:
            # Signal-to-noise estimate (using rolling std)
            rolling_std = gr_clean.rolling(10, min_periods=1).std()
            analysis['gr_snr_estimate'] = float(gr_clean.mean() / (rolling_std.mean() + 1e-8))

            # Autocorrelation at lag 1 (signal persistence)
            if len(gr_clean) > 20:
                analysis['gr_autocorr_lag1'] = float(gr_clean.autocorr(lag=1))

    return analysis

# Sample GR analysis
print("🔬 Gamma Ray Signal Analysis")
gr_analyses = []

for well_id in loader.train_well_ids[:50]:
    horizontal, typewell = loader.load_well_pair(well_id, 'train')
    gr_analysis = analyze_gr_signal(horizontal, typewell, well_id)
    gr_analyses.append(gr_analysis)

gr_summary = pd.DataFrame([{
    'well_id': a['well_id'],
    'h_gr_missing_pct': a.get('horizontal_gr', {}).get('missing_pct', np.nan),
    'h_gr_mean': a.get('horizontal_gr', {}).get('mean', np.nan),
    'h_gr_std': a.get('horizontal_gr', {}).get('std', np.nan),
    'h_gr_range': a.get('horizontal_gr', {}).get('max', np.nan) - a.get('horizontal_gr', {}).get('min', np.nan),
    'h_gr_snr': a.get('gr_snr_estimate', np.nan),
    'h_gr_autocorr': a.get('gr_autocorr_lag1', np.nan),
    't_gr_mean': a.get('typewell_gr', {}).get('mean', np.nan),
    't_gr_std': a.get('typewell_gr', {}).get('std', np.nan),
} for a in gr_analyses])

display(gr_summary.describe().round(2))

# GR distribution visualization
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Horizontal GR distribution (aggregated sample)
all_gr_h = []
for well_id in loader.train_well_ids[:30]:
    h = loader.load_horizontal(well_id, 'train')
    if 'GR' in h.columns:
        all_gr_h.extend(h['GR'].dropna().sample(min(100, len(h['GR'].dropna())), random_state=RANDOM_STATE).tolist())

if all_gr_h:
    axes[0, 0].hist(all_gr_h, bins=50, edgecolor='white', color=COLORS['gr'], alpha=0.7)
    axes[0, 0].set_xlabel('GR (API)')
    axes[0, 0].set_ylabel('Frequency')
    axes[0, 0].set_title('Horizontal Well GR Distribution (Sampled)')
    axes[0, 0].grid(alpha=0.3)

# GR missingness by well
missing_pcts = gr_summary['h_gr_missing_pct'].dropna()
if len(missing_pcts) > 0:
    axes[0, 1].hist(missing_pcts, bins=30, edgecolor='white', color=COLORS['warning'])
    axes[0, 1].axvline(missing_pcts.mean(), color='red', linestyle='--')
    axes[0, 1].set_xlabel('GR Missing Percentage')
    axes[0, 1].set_ylabel('Well Count')
    axes[0, 1].set_title('GR Missingness Distribution')
    axes[0, 1].grid(alpha=0.3)

# GR vs TVT scatter (sample well)
sample_well = loader.train_well_ids[0]
h_sample, t_sample = loader.load_well_pair(sample_well, 'train')
if 'GR' in h_sample.columns and 'TVT' in h_sample.columns:
    valid = h_sample[['GR', 'TVT']].dropna()
    if len(valid) > 0:
        axes[1, 0].scatter(valid['TVT'], valid['GR'], s=2, alpha=0.3, color=COLORS['primary'])
        axes[1, 0].set_xlabel('TVT (ft)')
        axes[1, 0].set_ylabel('GR (API)')
        axes[1, 0].set_title(f'GR vs TVT: {sample_well}')
        axes[1, 0].grid(alpha=0.3)

# GR autocorrelation distribution
autocorr_vals = gr_summary['h_gr_autocorr'].dropna()
if len(autocorr_vals) > 0:
    axes[1, 1].hist(autocorr_vals, bins=30, edgecolor='white', color=COLORS['success'])
    axes[1, 1].axvline(autocorr_vals.mean(), color='red', linestyle='--')
    axes[1, 1].set_xlabel('GR Autocorrelation (lag=1)')
    axes[1, 1].set_ylabel('Well Count')
    axes[1, 1].set_title('GR Signal Persistence')
    axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n💡 GR Insight: Mean autocorrelation = {gr_summary['h_gr_autocorr'].mean():.3f}")
print(f"   High persistence suggests GR is predictable along MD - useful for sequence models!")

# %%
# =============================================================================
# 7. GR CORRELATION: HORIZONTAL ↔ TYPEWELL (THE KEY FEATURE!)
# =============================================================================

def compute_gr_correlation(horizontal: pd.DataFrame, typewell: pd.DataFrame,
                         max_lag: int = 500) -> Dict:
    """
    Compute cross-correlation between horizontal and typewell GR signals
    This is THE most important feature engineering step for this competition
    """
    result = {}

    # Extract and clean GR data
    h_gr = horizontal['GR'].dropna().values if 'GR' in horizontal.columns else None
    h_tvt = horizontal['TVT_input'].dropna().values if 'TVT_input' in horizontal.columns else None
    t_gr = typewell['GR'].dropna().values if typewell is not None and 'GR' in typewell.columns else None
    t_tvt = typewell['TVT'].values if typewell is not None and 'TVT' in typewell.columns else None

    if h_gr is None or t_gr is None or len(h_gr) < 20 or len(t_gr) < 20:
        return {'error': 'Insufficient GR data'}

    # Normalize for correlation (z-score)
    h_gr_norm = (h_gr - h_gr.mean()) / (h_gr.std() + 1e-8)
    t_gr_norm = (t_gr - t_gr.mean()) / (t_gr.std() + 1e-8)

    # Cross-correlation to find optimal alignment
    correlation = correlate(h_gr_norm[:min(500, len(h_gr_norm))], t_gr_norm, mode='full')
    lags = np.arange(-len(t_gr_norm) + 1, len(h_gr_norm[:500]))

    best_lag_idx = np.argmax(correlation)
    best_lag = lags[best_lag_idx]
    max_correlation = correlation[best_lag_idx] / len(t_gr_norm)

    result['cross_correlation'] = {
        'best_lag': int(best_lag),
        'max_correlation': float(max_correlation),
        'correlation_at_zero': float(correlation[len(t_gr_norm)-1] / len(t_gr_norm)) if len(t_gr_norm) <= len(correlation) else None
    }

    # Map horizontal GR to typewell TVT scale using known TVT_input
    if h_tvt is not None and len(h_tvt) > 10 and t_tvt is not None:
        # Interpolate typewell GR onto TVT scale
        f_interp = interp1d(t_tvt, t_gr_norm, bounds_error=False, fill_value='extrapolate')

        # Use known TVT_input positions for mapping
        known_mask = horizontal['TVT_input'].notna()
        if known_mask.sum() > 10:
            h_known_tvt = horizontal.loc[known_mask, 'TVT_input'].values
            h_known_gr = horizontal.loc[known_mask, 'GR'].dropna().values

            if len(h_known_gr) > 10:
                # Map typewell GR to horizontal TVT positions
                mapped_t_gr = f_interp(h_known_tvt)

                # Normalize horizontal GR
                h_known_gr_norm = (h_known_gr - h_known_gr.mean()) / (h_known_gr.std() + 1e-8)

                # Compute similarity on overlapping region
                min_len = min(len(h_known_gr_norm), len(mapped_t_gr))
                if min_len > 10:
                    similarity = np.corrcoef(
                        h_known_gr_norm[:min_len],
                        mapped_t_gr[:min_len]
                    )[0, 1]
                    result['gr_similarity_known_zone'] = float(similarity)

    # Peak detection: do GR peaks align?
    h_peaks, _ = find_peaks(h_gr_norm, distance=10, prominence=0.5)
    t_peaks, _ = find_peaks(t_gr_norm, distance=10, prominence=0.5)

    result['peak_analysis'] = {
        'h_peaks_count': len(h_peaks),
        't_peaks_count': len(t_peaks),
        'peak_density_ratio': len(h_peaks) / (len(h_gr) + 1e-8) / (len(t_peaks) / (len(t_gr) + 1e-8))
    }

    return result

# Compute GR correlations for sample wells
print("🔗 GR Correlation Analysis: Horizontal ↔ Typewell")
correlation_results = []

for well_id in loader.train_well_ids[:30]:
    horizontal, typewell = loader.load_well_pair(well_id, 'train')
    if typewell is None:
        continue

    corr_result = compute_gr_correlation(horizontal, typewell)
    if 'error' not in corr_result:
        corr_result['well_id'] = well_id
        correlation_results.append(corr_result)

if correlation_results:
    corr_df = pd.DataFrame(correlation_results)

    print(f"\nGR Correlation Statistics:")
    if 'cross_correlation' in corr_df.columns:
        cc = corr_df['cross_correlation'].apply(lambda x: x.get('max_correlation') if isinstance(x, dict) else np.nan)
        print(f"  Mean max correlation: {cc.mean():.3f}")
        print(f"  Std max correlation: {cc.std():.3f}")
        print(f"  Min/Max: {cc.min():.3f} / {cc.max():.3f}")

    if 'gr_similarity_known_zone' in corr_df.columns:
        sim = corr_df['gr_similarity_known_zone'].dropna()
        if len(sim) > 0:
            print(f"\nGR Similarity in Known Zone:")
            print(f"  Mean: {sim.mean():.3f}")
            print(f"  Std: {sim.std():.3f}")

    # Visualization
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    if 'cross_correlation' in corr_df.columns:
        cc_vals = corr_df['cross_correlation'].apply(
            lambda x: x.get('max_correlation') if isinstance(x, dict) else np.nan
        ).dropna()

        axes[0].hist(cc_vals, bins=20, edgecolor='white', color=COLORS['accent'])
        axes[0].axvline(cc_vals.mean(), color='red', linestyle='--', linewidth=2)
        axes[0].set_xlabel('Max Cross-Correlation')
        axes[0].set_ylabel('Well Count')
        axes[0].set_title('GR Cross-Correlation Strength')
        axes[0].grid(alpha=0.3)

    if 'gr_similarity_known_zone' in corr_df.columns:
        sim_vals = corr_df['gr_similarity_known_zone'].dropna()
        if len(sim_vals) > 0:
            axes[1].scatter(range(len(sim_vals)), sim_vals, alpha=0.6, color=COLORS['primary'])
            axes[1].axhline(sim_vals.mean(), color='red', linestyle='--')
            axes[1].set_xlabel('Well Index')
            axes[1].set_ylabel('GR Similarity')
            axes[1].set_title('Known Zone GR Similarity')
            axes[1].grid(alpha=0.3)

    plt.tight_layout()
    plt.show()

    print(f"\n🎯 KEY INSIGHT: GR correlation strength varies significantly!")
    print(f"   Strong correlations (>0.7) indicate reliable typewell alignment.")
    print(f"   Weak correlations may require spatial/neighbor features as fallback.")

# %%
# =============================================================================
# 8. SPATIAL/TRAJECTORY ANALYSIS
# =============================================================================

def analyze_spatial_trajectory(horizontal: pd.DataFrame) -> Dict:
    """Analyze well trajectory in 3D space"""
    result = {}

    required_cols = ['MD', 'X', 'Y', 'Z']
    if not all(col in horizontal.columns for col in required_cols):
        return result

    # Basic spatial stats
    result['extent'] = {
        'X_range': float(horizontal['X'].max() - horizontal['X'].min()),
        'Y_range': float(horizontal['Y'].max() - horizontal['Y'].min()),
        'Z_range': float(horizontal['Z'].max() - horizontal['Z'].min()),
        'MD_range': float(horizontal['MD'].max() - horizontal['MD'].min())
    }

    # Directional derivatives
    dx = horizontal['X'].diff()
    dy = horizontal['Y'].diff()
    dz = horizontal['Z'].diff()
    dmd = horizontal['MD'].diff()

    # Heading (azimuth) and inclination
    heading = np.arctan2(dy, dx)
    inclination = np.arctan2(np.sqrt(dx**2 + dy**2), dz)

    result['directional_stats'] = {
        'mean_heading_rad': float(heading.dropna().mean()),
        'heading_std_rad': float(heading.dropna().std()),
        'mean_inclination_rad': float(inclination.dropna().mean()),
        'inclination_std_rad': float(inclination.dropna().std())
    }

    # Curvature estimate
    if len(horizontal) > 10:
        # Second derivatives for curvature
        d2x = dx.diff()
        d2y = dy.diff()
        d2z = dz.diff()

        curvature = np.sqrt(d2x**2 + d2y**2 + d2z**2) / (np.sqrt(dx**2 + dy**2 + dz**2)**3 + 1e-10)
        result['curvature'] = {
            'mean': float(curvature.dropna().mean()),
            'max': float(curvature.dropna().max()),
            'p95': float(curvature.dropna().quantile(0.95))
        }

    # Distance from wellhead
    x0, y0, z0 = horizontal.iloc[0][['X', 'Y', 'Z']]
    result['dist_from_head'] = {
        'mean': float(np.sqrt((horizontal['X']-x0)**2 + (horizontal['Y']-y0)**2 + (horizontal['Z']-z0)**2).mean()),
        'max': float(np.sqrt((horizontal['X']-x0)**2 + (horizontal['Y']-y0)**2 + (horizontal['Z']-z0)**2).max())
    }

    return result

# Spatial analysis across wells
print("🗺️ Spatial Trajectory Analysis")
spatial_stats = []

for well_id in loader.train_well_ids[:50]:
    horizontal = loader.load_horizontal(well_id, 'train')
    spatial = analyze_spatial_trajectory(horizontal)
    if spatial:
        spatial['well_id'] = well_id
        spatial_stats.append(spatial)

if spatial_stats:
    spatial_df = pd.DataFrame(spatial_stats)

    # Extract nested dicts
    extent_df = pd.json_normalize(spatial_df['extent'])
    directional_df = pd.json_normalize(spatial_df['directional_stats'])

    summary = pd.concat([
        spatial_df[['well_id']],
        extent_df.add_prefix('extent_'),
        directional_df.add_prefix('directional_')
    ], axis=1)

    display(summary.describe().round(2))

    # Map view visualization (sample wells)
    fig, ax = plt.subplots(figsize=(10, 10))

    # Plot all training wells (sampled for clarity)
    for well_id in loader.train_well_ids[:100]:
        h = loader.load_horizontal(well_id, 'train', usecols=['X', 'Y'])
        # Subsample for performance
        step = max(1, len(h) // 100)
        ax.plot(h['X'].iloc[::step], h['Y'].iloc[::step],
               color='lightgray', linewidth=0.5, alpha=0.3)

    # Highlight sample wells
    for well_id in loader.train_well_ids[:5]:
        h = loader.load_horizontal(well_id, 'train', usecols=['X', 'Y'])
        ax.plot(h['X'], h['Y'], label=well_id, linewidth=2)
        ax.scatter(h['X'].iloc[0], h['Y'].iloc[0], s=50, marker='*', color='red')  # Wellhead

    ax.set_xlabel('X / Easting (ft)')
    ax.set_ylabel('Y / Northing (ft)')
    ax.set_title('Well Trajectories: Map View (X-Y Plane)')
    ax.legend(loc='best', fontsize=8)
    ax.grid(alpha=0.3)
    ax.set_aspect('equal')
    plt.show()

    print(f"\n💡 Spatial Insight: Wells show varied trajectories!")
    print(f"   Neighboring wells may share geological dip - useful for spatial features.")

# %%
# =============================================================================
# 9. GEOLOGICAL FORMATION DEPTH ANALYSIS (FIXED)
# =============================================================================

GEOLOGICAL_COLS = ['ANCC', 'ASTNU', 'ASTNL', 'EGFDU', 'EGFDL', 'BUDA']

def analyze_geological_formations(horizontal: pd.DataFrame) -> Dict:
    """Analyze formation depth columns and their relationships - FIXED VERSION"""
    result = {}

    valid_cols = [c for c in GEOLOGICAL_COLS if c in horizontal.columns]
    if not valid_cols:
        return result

    # Basic stats per formation - ALWAYS include all expected keys
    formation_stats = {}
    for col in GEOLOGICAL_COLS:  # Iterate over ALL expected cols, not just present ones
        if col in horizontal.columns:
            valid = horizontal[col].dropna()
            if len(valid) > 0:
                formation_stats[col] = {
                    'count': len(valid),
                    'min': float(valid.min()),
                    'max': float(valid.max()),
                    'mean': float(valid.mean()),
                    'std': float(valid.std()),
                    'missing_pct': 100 * horizontal[col].isna().sum() / len(horizontal)
                }
            else:
                # Column exists but all NaN
                formation_stats[col] = {
                    'count': 0, 'min': np.nan, 'max': np.nan,
                    'mean': np.nan, 'std': np.nan,
                    'missing_pct': 100.0
                }
        else:
            # Column doesn't exist in this well
            formation_stats[col] = {
                'count': 0, 'min': np.nan, 'max': np.nan,
                'mean': np.nan, 'std': np.nan,
                'missing_pct': np.nan
            }
    result['formation_stats'] = formation_stats

    # Distance from Z to each formation (only if Z exists)
    if 'Z' in horizontal.columns:
        dist_features = {}
        for col in valid_cols:
            dist = horizontal['Z'] - horizontal[col]
            dist_features[f'dist_to_{col}'] = {
                'mean': float(dist.dropna().mean()) if dist.notna().any() else np.nan,
                'std': float(dist.dropna().std()) if dist.notna().any() else np.nan,
                'above_pct': 100 * (dist < 0).sum() / len(dist) if len(dist) > 0 else np.nan
            }
        result['dist_features'] = dist_features

    return result


# Analyze formation analysis across training wells (FIXED aggregation)
print("🪨 Geological Formation Analysis")
formation_records = []

for well_id in loader.train_well_ids[:30]:  # Sample for efficiency
    horizontal = loader.load_horizontal(well_id, 'train')
    analysis = analyze_geological_formations(horizontal)

    if not analysis or 'formation_stats' not in analysis:
        continue

    # Flatten the nested dict into a single row with consistent columns
    row = {'well_id': well_id}

    # Add formation stats with prefixed column names
    for col in GEOLOGICAL_COLS:
        stats = analysis['formation_stats'].get(col, {})
        for metric in ['mean', 'std', 'missing_pct']:
            row[f'{col}_{metric}'] = stats.get(metric, np.nan)

    # Add distance features if present
    if 'dist_features' in analysis:
        for feat_name, feat_stats in analysis['dist_features'].items():
            for metric in ['mean', 'above_pct']:
                row[f'{feat_name}_{metric}'] = feat_stats.get(metric, np.nan)

    formation_records.append(row)

# Now create DataFrame - all rows have same columns!
if formation_records:
    formation_df = pd.DataFrame(formation_records)

    print(f"\n✅ Successfully analyzed {len(formation_df)} wells")
    print(f"Columns: {len(formation_df.columns)}")

    # Display summary for key formations
    summary_cols = [c for c in formation_df.columns if any(f'{col}_mean' in c for col in GEOLOGICAL_COLS)]
    if summary_cols:
        print("\n📊 Formation Mean Depths (ft) Across Sampled Wells:")
        display(formation_df[['well_id'] + summary_cols].head(10).round(1))

    # Visualization: Box plot of formation depths
    plot_data = []
    for col in GEOLOGICAL_COLS:
        mean_col = f'{col}_mean'
        if mean_col in formation_df.columns:
            valid = formation_df[mean_col].dropna()
            if len(valid) > 5:  # Need minimum wells to plot
                for val in valid:
                    plot_data.append({'formation': col, 'depth_ft': val})

    if plot_data:
        plot_df = pd.DataFrame(plot_data)

        plt.figure(figsize=(10, 6))
        sns.boxplot(data=plot_df, x='formation', y='depth_ft', palette='Set2')
        plt.xlabel('Geological Formation')
        plt.ylabel('Depth (ft, Z coordinate)')
        plt.title('Formation Depth Distribution Across Wells')
        plt.xticks(rotation=45, ha='right')
        plt.grid(alpha=0.3, axis='y')
        plt.tight_layout()
        plt.show()

        print(f"\n💡 Formation Insight: Depths are relatively consistent within formations!")
        print(f"   Distance-to-formation features (Z - formation_depth) will be powerful predictors.")
else:
    print("⚠️ No formation data found in sampled wells")

# %%
# =============================================================================
# 10. MISSINGNESS & DATA QUALITY AUDIT
# =============================================================================

def comprehensive_missingness_analysis(data_root: Path, split: str = 'train',
                                      max_wells: int = 100) -> pd.DataFrame:
    """Analyze missing data patterns across the dataset"""
    directory = data_root / split
    missing_summary = []

    for i, filepath in enumerate(directory.glob('*__horizontal_well.csv')):
        if i >= max_wells:
            break

        df = pd.read_csv(filepath)
        well_id = filepath.stem.split('__')[0]

        for col in df.columns:
            missing_count = df[col].isna().sum()
            missing_summary.append({
                'well_id': well_id,
                'column': col,
                'missing_count': int(missing_count),
                'total_rows': len(df),
                'missing_pct': 100 * missing_count / len(df),
                'dtype': str(df[col].dtype)
            })

    return pd.DataFrame(missing_summary)

print("🔍 Missingness Analysis")
missing_df = comprehensive_missingness_analysis(DATA_ROOT, 'train', max_wells=50)

# Aggregate by column
column_missingness = missing_df.groupby('column').agg({
    'missing_pct': ['mean', 'std', 'min', 'max'],
    'well_id': 'count'
}).round(2)
column_missingness.columns = ['_'.join(col).strip() for col in column_missingness.columns]
column_missingness = column_missingness.sort_values('missing_pct_mean', ascending=False)

print("\nMissingness by Column (Training Set):")
display(column_missingness.head(15))

# Missingness heatmap for sample wells
sample_wells_missing = missing_df[missing_df['well_id'].isin(loader.train_well_ids[:10])]
pivot_missing = sample_wells_missing.pivot_table(
    index='well_id',
    columns='column',
    values='missing_pct',
    aggfunc='mean'
).fillna(0)

if not pivot_missing.empty:
    plt.figure(figsize=(12, 6))
    sns.heatmap(pivot_missing, annot=True, fmt='.1f', cmap='YlOrRd', center=50)
    plt.title('Missingness Heatmap: Sample Wells')
    plt.xlabel('Column')
    plt.ylabel('Well ID')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

print(f"\n💡 Missingness Insight:")
print(f"   - TVT_input: ~74% missing by design (prediction zone)")
print(f"   - GR: ~30% missing - consider imputation or masking strategies")
print(f"   - Formation columns: Low missingness, good for features")

# %%
# =============================================================================
# 11. FEATURE CORRELATION & MULTICOLLINEARITY ANALYSIS
# =============================================================================

def sample_correlation_analysis(data_root: Path, split: str = 'train',
                               sample_per_well: int = 200, max_wells: int = 30) -> pd.DataFrame:
    """Sample rows from multiple wells to compute feature correlations"""
    directory = data_root / split
    sampled_data = []

    numeric_cols = ['MD', 'X', 'Y', 'Z', 'GR', 'TVT', 'TVT_input'] + GEOLOGICAL_COLS

    for i, filepath in enumerate(directory.glob('*__horizontal_well.csv')):
        if i >= max_wells:
            break

        df = pd.read_csv(filepath, usecols=[c for c in numeric_cols if c in pd.read_csv(filepath, nrows=0).columns])
        existing_cols = [c for c in numeric_cols if c in df.columns]

        if len(df) > sample_per_well:
            df = df.sample(sample_per_well, random_state=RANDOM_STATE)

        sampled_data.append(df[existing_cols])

    if not sampled_data:
        return pd.DataFrame()

    combined = pd.concat(sampled_data, ignore_index=True)
    return combined.select_dtypes(include=[np.number]).corr()

print("🔗 Feature Correlation Analysis")
corr_matrix = sample_correlation_analysis(DATA_ROOT, 'train')

if not corr_matrix.empty and corr_matrix.shape[0] > 1:
    # Focus on TVT correlations (target variable)
    if 'TVT' in corr_matrix.columns:
        tvt_corr = corr_matrix['TVT'].drop('TVT').sort_values(key=abs, ascending=False)
        print("\nTop Correlations with TVT (Target):")
        display(tvt_corr.head(10).round(3))

    # Multicollinearity check (VIF approximation via correlation)
    high_corr_pairs = []
    cols = corr_matrix.columns.tolist()
    for i, col1 in enumerate(cols[:-1]):
        for col2 in cols[i+1:]:
            corr_val = abs(corr_matrix.loc[col1, col2])
            if corr_val > 0.9 and col1 != col2:
                high_corr_pairs.append((col1, col2, corr_val))

    if high_corr_pairs:
        print(f"\n⚠️ High Multicollinearity Pairs (|r| > 0.9):")
        for c1, c2, r in high_corr_pairs[:10]:
            print(f"   {c1} ↔ {c2}: {r:.3f}")

    # Correlation heatmap
    plt.figure(figsize=(14, 12))
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
    sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
                center=0, square=True, linewidths=0.5)
    plt.title('Feature Correlation Matrix (Sampled)')
    plt.tight_layout()
    plt.show()

    print(f"\n💡 Correlation Insight:")
    print(f"   - TVT_input highly correlated with TVT (expected in known zone)")
    print(f"   - GR shows moderate correlation with TVT - useful but not deterministic")
    print(f"   - Spatial coords (X,Y,Z) have complex relationships - consider interactions")

# %%
# =============================================================================
# 12. NEIGHBORING WELL CONTEXT ANALYSIS
# =============================================================================

def find_neighboring_wells(well_id: str, train_well_ids: List[str],
                          loader: WellboreDataLoader,
                          max_distance_ft: float = 10000,
                          max_neighbors: int = 10) -> List[Dict]:
    """Find spatially neighboring wells based on wellhead coordinates"""
    # Get target well's wellhead
    target_h = loader.load_horizontal(well_id, 'train', usecols=['X', 'Y', 'Z']).iloc[0]
    target_head = np.array([target_h['X'], target_h['Y'], target_h['Z']])

    neighbors = []

    for other_id in train_well_ids:
        if other_id == well_id:
            continue

        try:
            other_h = loader.load_horizontal(other_id, 'train', usecols=['X', 'Y', 'Z']).iloc[0]
            other_head = np.array([other_h['X'], other_h['Y'], other_h['Z']])

            distance = np.linalg.norm(target_head - other_head)

            if distance <= max_distance_ft:
                neighbors.append({
                    'well_id': other_id,
                    'distance_ft': float(distance),
                    'dx': float(other_head[0] - target_head[0]),
                    'dy': float(other_head[1] - target_head[1]),
                    'dz': float(other_head[2] - target_head[2])
                })
        except:
            continue

    # Sort by distance and return top neighbors
    neighbors.sort(key=lambda x: x['distance_ft'])
    return neighbors[:max_neighbors]

print("👥 Neighboring Well Analysis")
# Analyze neighbors for sample wells
neighbor_stats = []

for well_id in loader.train_well_ids[:20]:
    neighbors = find_neighboring_wells(well_id, loader.train_well_ids, loader)

    neighbor_stats.append({
        'well_id': well_id,
        'n_neighbors_5kft': sum(1 for n in neighbors if n['distance_ft'] <= 5000),
        'n_neighbors_10kft': len(neighbors),
        'min_neighbor_distance': neighbors[0]['distance_ft'] if neighbors else np.inf,
        'mean_neighbor_distance': np.mean([n['distance_ft'] for n in neighbors]) if neighbors else np.inf
    })

neighbor_df = pd.DataFrame(neighbor_stats)
print("\nNeighboring Well Statistics:")
display(neighbor_df.describe().round(1))

# Visualization: Neighbor distance distribution
if neighbor_df['min_neighbor_distance'].notna().any():
    plt.figure(figsize=(10, 5))
    valid_distances = neighbor_df['min_neighbor_distance'][neighbor_df['min_neighbor_distance'] < np.inf]
    plt.hist(valid_distances, bins=30, edgecolor='white', color=COLORS['spatial'])
    plt.axvline(valid_distances.median(), color='red', linestyle='--', label=f'Median: {valid_distances.median():.0f} ft')
    plt.xlabel('Distance to Nearest Neighbor (ft)')
    plt.ylabel('Well Count')
    plt.title('Spatial Proximity: Distance to Nearest Training Well')
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()

print(f"\n💡 Neighbor Insight: Most wells have neighbors within ~5,000 ft!")
print(f"   Spatial features from neighboring wells could improve predictions.")
print(f"   Consider: average TVT gradient, GR signature similarity, formation depth offsets")

# %%
# =============================================================================
# 13. KEY INSIGHTS SUMMARY FOR MODELING
# =============================================================================

print("\n" + "="*80)
print("🎯 KEY INSIGHTS SUMMARY: ACTIONABLE RECOMMENDATIONS FOR MODELING")
print("="*80)

insights = [
    ("📊 Target Variable (TVT)", [
        "TVT is NOT monotonic: ~48% of segments are 'nearly flat', 33% increasing, 19% decreasing",
        "Models must handle bidirectional TVT changes - avoid monotonicity constraints",
        "TVT range per well: ~300-1400 ft - consider normalization per well"
    ]),

    ("🔑 Critical Feature: GR Correlation", [
        "Cross-correlate horizontal GR with typewell GR to align on TVT scale",
        "Correlation strength varies (0.3-0.9) - use as confidence weight",
        "Known TVT_input zone provides ground truth for calibration"
    ]),

    ("📈 Sequence Modeling Opportunity", [
        "GR shows high autocorrelation (mean ~0.85) - predictable along MD",
        "TVT derivatives (dTVT/dMD) provide trend signals for prediction",
        "Use LSTM/Transformer to capture long-range dependencies along wellbore"
    ]),

    ("🗺️ Spatial Context Features", [
        "Most wells have neighbors within 5,000 ft - leverage offset-well data",
        "Distance-to-formation features (Z - formation_depth) are powerful",
        "Well trajectory (heading, inclination) may correlate with geological dip"
    ]),

    ("🛡️ Validation Strategy", [
        "ALWAYS use well-level GroupKFold - never split rows from same well",
        "Mimic test setup: hide evaluation zone in training for OOF validation",
        "Track OOF RMSE per well to identify difficult cases"
    ]),

    ("⚠️ Data Quality Considerations", [
        "TVT_input: 74% missing by design - this IS the prediction zone",
        "GR: ~30% missing - use forward-fill or model-based imputation",
        "Formation columns: Low missingness, excellent for features"
    ]),

    ("🚀 Modeling Recommendations", [
        "Baseline: LightGBM with derivative + spatial + GR-correlation features",
        "Advanced: Ensemble with Transformer for sequence modeling",
        "Post-processing: Apply rolling smooth to predictions along MD",
        "Ensemble weights: Favor models with strong GR-correlation features"
    ])
]

for title, points in insights:
    print(f"\n{title}")
    print("-" * len(title))
    for point in points:
        print(f"  • {point}")

print("\n" + "="*80)
print("✅ EDA Complete! Ready for feature engineering and modeling.")
print("="*80)
```
